# Cohort Retention Analysis — SaaS Churn & Revenue

1. Load/generate subscription data
2. Assign cohorts and compute retention matrix
3. Visualize retention heatmap
4. Export for Power BI (ARR dashboard)

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path("../scripts").resolve()))

import pandas as pd
from etl_subscription_logs import run_etl
from cohort_analysis import assign_cohort, compute_retention_matrix, pivot_retention

data_dir = Path("../data").resolve()
subs, customers = run_etl(data_dir)
print(subs.shape, customers.shape)
```

```python
subs["period_month"] = subs["start_date"].dt.to_period("M")
subs = assign_cohort(subs)
subs["active"] = subs["end_date"].isna().astype(int)

retention_agg = compute_retention_matrix(subs)
matrix = pivot_retention(retention_agg)
matrix
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))
sns.heatmap(matrix, annot=True, fmt=".0f", cmap="RdYlGn", vmin=0, vmax=100)
plt.title("Cohort Retention (%)")
plt.tight_layout()
plt.show()
```

```python
matrix.to_csv(data_dir / "cohort_retention_matrix.csv")
print("Exported cohort_retention_matrix.csv for Power BI")
```
